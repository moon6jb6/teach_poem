```python
# 运行前请把 YOUR_ARK_API_KEY 换成自己的火山方舟（豆包）API Key；Key 不要提交到仓库。
from flask import Flask, request, jsonify
from flask_cors import CORS
from openai import OpenAI
CORS((app:=Flask(__name__)))
from volcenginesdkarkruntime import Ark

client = Ark(base_url='https://ark.cn-beijing.volces.com/api/v3',api_key="YOUR_ARK_API_KEY")
l = []
@app.route("/api/chat_text",methods=["POST"])
def chat_t():
    data = (request.get_json() or {}).get("a1", "").strip()
    
    if (i := (request.get_json() or {}).get("a0", "")) not in l:
        response = client.responses.create(
            model="doubao-seed-2-0-lite-260215",
            input = i)
        l.append(i)
        
        otpt = response.output[1].content[0].text
        print(otpt,'first')
        return jsonify({
        "success": True,
        "reply":'  '+otpt
    })
    response = client.responses.create(
        model="doubao-seed-2-0-lite-260215",
        input = data)
    otpt = response.output[1].content[0].text
    print(otpt)
    
    return jsonify({
        "success": True,
        "reply": '  '+otpt
        # "reply": data+'  '+otpt
    })



if __name__ == "__main__":
    app.run(host="127.0.0.1", port=5000, debug=True,use_reloader=False)
```
